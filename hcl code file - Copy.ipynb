```python

class Parcel:
    def __init__(self, parcel_id, sender, receiver, address):
        self.parcel_id = parcel_id
        self.sender = sender
        self.receiver = receiver
        self.address = address
        self.status = "Booked"


class CourierService:

    def __init__(self):
        self.parcels = {}

    def create_parcel(self):
        parcel_id = input("Enter parcel ID: ")
        sender = input("Enter sender name: ")
        receiver = input("Enter receiver name: ")
        address = input("Enter delivery address: ")

        parcel = Parcel(parcel_id, sender, receiver, address)
        self.parcels[parcel_id] = parcel

        print("Parcel registered successfully!")

    def update_status(self):
        parcel_id = input("Enter parcel ID: ")

        if parcel_id in self.parcels:
            new_status = input("Enter new status: ")
            self.parcels[parcel_id].status = new_status
            print("Parcel status updated successfully!")
        else:
            print("Parcel not found!")

    def track_parcel(self):
        parcel_id = input("Enter parcel ID to track: ")

        if parcel_id in self.parcels:
            parcel = self.parcels[parcel_id]

            print("\n--- Parcel Details ---")
            print("Parcel ID:", parcel.parcel_id)
            print("Sender:", parcel.sender)
            print("Receiver:", parcel.receiver)
            print("Address:", parcel.address)
            print("Status:", parcel.status)
        else:
            print("Parcel not found!")


# Creating CourierService object
courier = CourierService()

while True:
    print("\n===== COURIER TRACKING SYSTEM =====")
    print("1. Register Parcel")
    print("2. Update Parcel Status")
    print("3. Track Parcel")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        courier.create_parcel()

    elif choice == "2":
        courier.update_status()

    elif choice == "3":
        courier.track_parcel()

    elif choice == "4":
        print("Thank you for using Courier Tracking System!")
        break

    else:
        print("Invalid choice! Please try again.")
```
